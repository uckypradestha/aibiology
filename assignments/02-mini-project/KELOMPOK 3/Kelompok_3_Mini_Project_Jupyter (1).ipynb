```python

# ============================================================
# ANALISIS DATA PERTUMBUHAN Gracilaria verrucosa
# REGRESI, KLASIFIKASI, PCA, DAN K-MEANS
# ============================================================

# Jika menggunakan Google Colab dan library belum tersedia:
# !pip install pandas numpy matplotlib seaborn scikit-learn openpyxl

# ============================================================
# 1. IMPORT LIBRARY
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import GroupShuffleSplit
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import (
    RandomForestRegressor,
    RandomForestClassifier
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")


# ============================================================
# 2. MEMBACA DATA EXCEL
# ============================================================

FILE_EXCEL = "Dataset_Gracilaria_Sederhana.xlsx"

df = pd.read_excel(
    FILE_EXCEL,
    sheet_name="Data Pertumbuhan"
)

sumber = pd.read_excel(
    FILE_EXCEL,
    sheet_name="Daftar Jurnal"
)

# Menghapus baris template yang belum diisi
df = df.dropna(
    subset=["Kode Jurnal"]
).copy()

print("Jumlah data awal:", len(df))


# ============================================================
# 3. FILTER SPESIES
# ============================================================

# Penelitian dibatasi pada Gracilaria verrucosa

df["Jenis Gracilaria"] = (
    df["Jenis Gracilaria"]
    .astype(str)
    .str.strip()
)

df = df[
    df["Jenis Gracilaria"]
    .str.lower()
    .eq("gracilaria verrucosa")
].copy()

print("Jumlah data Gracilaria verrucosa:", len(df))

if len(df) == 0:
    raise ValueError(
        "Tidak ditemukan data Gracilaria verrucosa. "
        "Periksa kembali penulisan nama spesies di Excel."
    )


# ============================================================
# 4. PEMBERSIHAN DAN PEMERIKSAAN DATA
# ============================================================

kolom_angka = [
    "Tahun Jurnal",
    "Lama Budidaya (hari)",
    "Hari Pengamatan",
    "Berat Awal (gram)",
    "Berat Akhir (gram)",
    "Suhu Air (°C)",
    "pH Air",
    "Salinitas (ppt)",
    "Oksigen Terlarut (mg/L)",
    "Nitrat (mg/L)",
    "Fosfat (mg/L)",
    "Laju Pertumbuhan (%/hari)"
]

for kolom in kolom_angka:
    if kolom in df.columns:
        df[kolom] = pd.to_numeric(
            df[kolom],
            errors="coerce"
        )

# Menghapus duplikasi yang benar-benar identik
df = df.drop_duplicates().copy()

print("\nUkuran dataset:", df.shape)

print("\nTipe data:")
display(df.dtypes)

print("\nJumlah data kosong:")
display(df.isna().sum())

print("\nJumlah data duplikat:", df.duplicated().sum())

print("\nRingkasan data:")
display(df.describe(include="all").T)


# ============================================================
# 5. PERHITUNGAN SGR (SEBAGAI INFORMASI TAMBAHAN)
# ============================================================

# SGR tidak menjadi target utama penelitian ini.
# Perhitungan hanya dilakukan jika berat awal dan akhir
# menggunakan basis yang sama dan lama pengamatan tersedia.

syarat_sgr = (
    df["Berat Awal (gram)"].gt(0) &
    df["Berat Akhir (gram)"].gt(0) &
    df["Hari Pengamatan"].gt(0)
)

hitung_sgr = (
    syarat_sgr &
    df["Laju Pertumbuhan (%/hari)"].isna()
)

df.loc[hitung_sgr, "Laju Pertumbuhan (%/hari)"] = (
    (
        np.log(df.loc[hitung_sgr, "Berat Akhir (gram)"]) -
        np.log(df.loc[hitung_sgr, "Berat Awal (gram)"])
    ) /
    df.loc[hitung_sgr, "Hari Pengamatan"]
) * 100

print("\nData SGR sebagai informasi tambahan:")
display(
    df[
        [
            "Kode Jurnal",
            "Berat Awal (gram)",
            "Berat Akhir (gram)",
            "Hari Pengamatan",
            "Laju Pertumbuhan (%/hari)"
        ]
    ].head()
)


# ============================================================
# 6. EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# Statistik deskriptif
display(df.describe(include="all").T)

# Distribusi variabel
kolom_plot = [
    "Suhu Air (°C)",
    "pH Air",
    "Salinitas (ppt)",
    "Oksigen Terlarut (mg/L)",
    "Berat Awal (gram)",
    "Berat Akhir (gram)",
    "Laju Pertumbuhan (%/hari)"
]

tersedia = [
    kolom for kolom in kolom_plot
    if kolom in df.columns and df[kolom].notna().any()
]

for kolom in tersedia:

    plt.figure(figsize=(7, 4))

    sns.histplot(
        df[kolom].dropna(),
        kde=True
    )

    plt.title(f"Distribusi {kolom}")
    plt.xlabel(kolom)
    plt.ylabel("Frekuensi")

    plt.tight_layout()
    plt.show()


# ============================================================
# 7. ANALISIS KORELASI
# ============================================================

numerik = df.select_dtypes(include=np.number)

if numerik.shape[1] >= 2:

    plt.figure(figsize=(12, 8))

    sns.heatmap(
        numerik.corr(),
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Korelasi Antarvariabel Gracilaria verrucosa")
    plt.tight_layout()
    plt.show()

else:
    print("Data numerik belum cukup untuk analisis korelasi.")


# ============================================================
# 8. SUPERVISED LEARNING: REGRESI
# ============================================================

# Target regresi:
# Biomassa akhir (gram)

target_regresi = "Berat Akhir (gram)"

# Fitur prediktor mengikuti struktur kode awal.
# Pertambahan berat tidak dimasukkan untuk menghindari
# target leakage jika biomassa akhir dihitung dari berat awal
# + pertambahan berat.

fitur_regresi = [
    "Suhu Air (°C)",
    "pH Air",
    "Salinitas (ppt)",
    "Oksigen Terlarut (mg/L)",
    "Nitrat (mg/L)",
    "Fosfat (mg/L)",
    "Lama Budidaya (hari)",
    "Hari Pengamatan",
    "Berat Awal (gram)"
]

reg = df.dropna(
    subset=[target_regresi, "Kode Jurnal"]
).copy()

fitur_regresi = [
    kolom for kolom in fitur_regresi
    if kolom in reg.columns
    and reg[kolom].notna().any()
]

print("\nJumlah data untuk regresi:", len(reg))
print("Jumlah jurnal independen:", reg["Kode Jurnal"].nunique())
print("Fitur regresi:", fitur_regresi)


if (
    len(reg) >= 10
    and reg["Kode Jurnal"].nunique() >= 2
    and len(fitur_regresi) >= 2
):

    X = reg[fitur_regresi]
    y = reg[target_regresi]
    groups = reg["Kode Jurnal"]

    # Pembagian berdasarkan jurnal
    splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=0.25,
        random_state=42
    )

    train_idx, test_idx = next(
        splitter.split(X, y, groups=groups)
    )

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    print("\nJumlah data latih:", len(X_train))
    print("Jumlah data uji:", len(X_test))

    # Model Random Forest Regressor
    model_reg = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42
            )
        )
    ])

    model_reg.fit(X_train, y_train)

    pred_reg = model_reg.predict(X_test)

    # Evaluasi regresi
    mae = mean_absolute_error(y_test, pred_reg)

    rmse = np.sqrt(
        mean_squared_error(y_test, pred_reg)
    )

    r2 = r2_score(y_test, pred_reg)

    print("\n===== HASIL REGRESI =====")
    print("MAE :", mae)
    print("RMSE:", rmse)
    print("R2  :", r2)

    # Perbandingan nilai aktual dan prediksi
    hasil_regresi = pd.DataFrame({
        "Biomassa Aktual": y_test.values,
        "Biomassa Prediksi": pred_reg
    })

    display(hasil_regresi)

    # Grafik aktual vs prediksi
    plt.figure(figsize=(7, 5))

    plt.scatter(
        y_test,
        pred_reg
    )

    nilai_min = min(y_test.min(), pred_reg.min())
    nilai_max = max(y_test.max(), pred_reg.max())

    plt.plot(
        [nilai_min, nilai_max],
        [nilai_min, nilai_max],
        linestyle="--"
    )

    plt.xlabel("Biomassa Aktual (gram)")
    plt.ylabel("Biomassa Prediksi (gram)")
    plt.title("Random Forest Regression - Biomassa Gracilaria")

    plt.tight_layout()
    plt.show()

    # Feature importance
    rf_reg = model_reg.named_steps["model"]

    importance = pd.DataFrame({
        "Fitur": fitur_regresi,
        "Importance": rf_reg.feature_importances_
    }).sort_values(
        by="Importance",
        ascending=False
    )

    print("\nKontribusi fitur regresi:")
    display(importance)

    plt.figure(figsize=(8, 5))

    sns.barplot(
        data=importance,
        x="Importance",
        y="Fitur"
    )

    plt.title("Feature Importance Random Forest Regression")
    plt.tight_layout()
    plt.show()

else:

    print(
        "Regresi dilewati: diperlukan minimal 10 data, "
        "2 jurnal independen, dan 2 fitur yang tersedia."
    )


# ============================================================
# 9. SUPERVISED LEARNING: KLASIFIKASI
# ============================================================

# Klasifikasi tetap dipertahankan sebagai analisis eksploratif.
# Kategori dibuat berdasarkan distribusi biomassa:
# Rendah, Sedang, dan Tinggi.
#
# Kategori ini bukan standar biologis universal.

target_klasifikasi = "Berat Akhir (gram)"

klas = df.dropna(
    subset=[target_klasifikasi, "Kode Jurnal"]
).copy()

fitur_klas = [
    kolom for kolom in fitur_regresi
    if kolom in klas.columns
]

if (
    len(klas) >= 15
    and klas["Kode Jurnal"].nunique() >= 2
    and len(fitur_klas) >= 2
):

    X = klas[fitur_klas]
    y_cont = klas[target_klasifikasi]
    groups = klas["Kode Jurnal"]

    splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=0.25,
        random_state=42
    )

    train_idx, test_idx = next(
        splitter.split(X, y_cont, groups=groups)
    )

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train_cont = y_cont.iloc[train_idx]
    y_test_cont = y_cont.iloc[test_idx]

    # Ambang kelas hanya dihitung dari data latih
    batas = y_train_cont.quantile(
        [1/3, 2/3]
    ).values

    # Periksa apakah ambang batas menghasilkan 3 kelas berbeda
    if batas[0] < batas[1]:

        def buat_kategori(nilai):
            return pd.cut(
                nilai,
                bins=[
                    -np.inf,
                    batas[0],
                    batas[1],
                    np.inf
                ],
                labels=[
                    "Rendah",
                    "Sedang",
                    "Tinggi"
                ],
                include_lowest=True
            )

        y_train = buat_kategori(y_train_cont)
        y_test = buat_kategori(y_test_cont)

        if y_train.nunique() >= 2:

            # Model Random Forest Classifier
            model_klas = Pipeline([
                (
                    "imputer",
                    SimpleImputer(strategy="median")
                ),
                (
                    "scaler",
                    StandardScaler()
                ),
                (
                    "model",
                    RandomForestClassifier(
                        n_estimators=200,
                        random_state=42,
                        class_weight="balanced"
                    )
                )
            ])

            model_klas.fit(
                X_train,
                y_train
            )

            pred_klas = model_klas.predict(
                X_test
            )

            print("\n===== HASIL KLASIFIKASI =====")

            print(
                "Akurasi:",
                accuracy_score(y_test, pred_klas)
            )

            print(
                "F1 Macro:",
                f1_score(
                    y_test,
                    pred_klas,
                    average="macro",
                    zero_division=0
                )
            )

            print("\nClassification Report:")

            print(
                classification_report(
                    y_test,
                    pred_klas,
                    zero_division=0
                )
            )

            # Confusion Matrix
            label_kelas = [
                "Rendah",
                "Sedang",
                "Tinggi"
            ]

            cm = confusion_matrix(
                y_test,
                pred_klas,
                labels=label_kelas
            )

            disp = ConfusionMatrixDisplay(
                confusion_matrix=cm,
                display_labels=label_kelas
            )

            disp.plot(
                cmap="Blues",
                values_format="d"
            )

            plt.title("Confusion Matrix Random Forest Classifier")
            plt.tight_layout()
            plt.show()

            print("Batas kategori biomassa:", batas)

        else:

            print(
                "Klasifikasi dilewati: "
                "data latih hanya memiliki satu kategori."
            )

    else:

        print(
            "Klasifikasi dilewati: "
            "nilai biomassa tidak menghasilkan batas kelas yang berbeda."
        )

else:

    print(
        "Klasifikasi dilewati: data belum mencukupi "
        "untuk analisis eksploratif."
    )


# ============================================================
# 10. UNSUPERVISED LEARNING: PCA DAN K-MEANS
# ============================================================

# Analisis menggunakan variabel kualitas air.
# Tujuannya mengelompokkan kemiripan kondisi budidaya.

fitur_cluster = [
    "Suhu Air (°C)",
    "pH Air",
    "Salinitas (ppt)",
    "Oksigen Terlarut (mg/L)",
    "Nitrat (mg/L)",
    "Fosfat (mg/L)"
]

fitur_cluster = [
    kolom for kolom in fitur_cluster
    if kolom in df.columns
    and df[kolom].notna().any()
]

clu = df.dropna(
    subset=["Kode Jurnal"]
).copy()

# Buang baris yang seluruh variabel clustering-nya kosong
clu = clu.dropna(
    subset=fitur_cluster,
    how="all"
)

if (
    len(fitur_cluster) >= 2
    and len(clu) >= 6
):

    Xc = clu[fitur_cluster]

    # Imputasi dan standardisasi
    prep = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    Xc_scaled = prep.fit_transform(Xc)

    # PCA
    pca = PCA(
        n_components=2,
        random_state=42
    )

    pcs = pca.fit_transform(
        Xc_scaled
    )

    print("\n===== HASIL PCA =====")

    print(
        "Proporsi variasi komponen:",
        pca.explained_variance_ratio_
    )

    print(
        "Total variasi yang dijelaskan:",
        pca.explained_variance_ratio_.sum()
    )

    # Menentukan jumlah cluster menggunakan silhouette score
    skor_silhouette = {}

    batas_cluster = min(
        6,
        len(clu) - 1
    )

    for k in range(2, batas_cluster + 1):

        km_uji = KMeans(
            n_clusters=k,
            random_state=42,
            n_init=10
        )

        label_uji = km_uji.fit_predict(
            Xc_scaled
        )

        if len(np.unique(label_uji)) > 1:

            skor = silhouette_score(
                Xc_scaled,
                label_uji
            )

            skor_silhouette[k] = skor

    if len(skor_silhouette) > 0:

        print("\nSilhouette Score setiap cluster:")

        for k, skor in skor_silhouette.items():
            print(f"{k} Cluster: {skor:.4f}")

        # Memilih jumlah cluster dengan silhouette score tertinggi
        jumlah_cluster = max(
            skor_silhouette,
            key=skor_silhouette.get
        )

        print(
            "\nJumlah cluster terbaik:",
            jumlah_cluster
        )

        # K-Means final
        km = KMeans(
            n_clusters=jumlah_cluster,
            random_state=42,
            n_init=10
        )

        labels = km.fit_predict(
            Xc_scaled
        )

        clu["Cluster"] = labels

        # Visualisasi PCA dan K-Means
        plt.figure(figsize=(8, 6))

        plt.scatter(
            pcs[:, 0],
            pcs[:, 1],
            c=labels
        )

        plt.xlabel("Komponen Utama 1")
        plt.ylabel("Komponen Utama 2")

        plt.title(
            "PCA dan K-Means Gracilaria verrucosa"
        )

        plt.tight_layout()
        plt.show()

        print("\nHasil pengelompokan:")
        display(
            clu[
                ["Kode Jurnal", "Cluster"] +
                fitur_cluster
            ]
        )

    else:

        print(
            "K-Means dilewati: "
            "cluster tidak dapat dibentuk secara valid."
        )

else:

    print(
        "PCA/K-Means dilewati: diperlukan minimal "
        "2 fitur kualitas air dan 6 baris data."
    )


# ============================================================
# 11. MENYIMPAN DATA HASIL PEMBERSIHAN
# ============================================================

file_output = "Data_Gracilaria_verrucosa_Bersih.xlsx"

df.to_excel(
    file_output,
    index=False
)

print("\nDataset hasil pembersihan berhasil disimpan:")
print(file_output)


# ============================================================
# 12. CATATAN INTERPRETASI
# ============================================================

print("""
CATATAN ANALISIS:

1. Regresi digunakan untuk memperkirakan biomassa akhir
   Gracilaria verrucosa.

2. Klasifikasi biomassa rendah, sedang, dan tinggi
   merupakan pengelompokan relatif berdasarkan dataset,
   bukan standar biologis universal.

3. PCA dan K-Means digunakan untuk melihat pola kemiripan
   kondisi kualitas air selama budidaya.

4. Data dari satu jurnal tidak boleh tersebar ke data
   latih dan data uji.

5. Biomassa basah dan kering tidak boleh digabungkan
   tanpa dasar konversi yang valid.

6. Jika jumlah jurnal independen masih sedikit,
   hasil model hanya bersifat eksploratif dan belum
   dapat digeneralisasikan.
""")


hasil_klasifikasi = pd.DataFrame({
    "Aktual": y_test.values,
    "Prediksi": pred_klas
})

print("Perbandingan kategori aktual dan prediksi:")
display(hasil_klasifikasi)

print("\nConfusion Matrix:")
from sklearn.metrics import confusion_matrix

print(confusion_matrix(
    y_test,
    pred_klas,
    labels=["Rendah", "Sedang", "Tinggi"]
))

print("Jumlah data latih:")
print(y_train.value_counts())

print("\nJumlah data uji:")
print(y_test.value_counts())

print("\nDistribusi prediksi:")
print(pd.Series(pred_klas).value_counts())

print("\nJumlah jurnal independen:")
print(klas["Kode Jurnal"].nunique())
```

```python

# ============================================================
# ANALISIS DATA PERTUMBUHAN Gracilaria verrucosa
# REGRESI, KLASIFIKASI, PCA, DAN K-MEANS
# ============================================================

# Jika menggunakan Google Colab dan library belum tersedia:
# !pip install pandas numpy matplotlib seaborn scikit-learn openpyxl

# ============================================================
# 1. IMPORT LIBRARY
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import GroupShuffleSplit
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import (
    RandomForestRegressor,
    RandomForestClassifier
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")


# ============================================================
# 2. MEMBACA DATA EXCEL
# ============================================================

FILE_EXCEL = "Dataset_Gracilaria_Sederhana.xlsx"

df = pd.read_excel(
    FILE_EXCEL,
    sheet_name="Data Pertumbuhan"
)

sumber = pd.read_excel(
    FILE_EXCEL,
    sheet_name="Daftar Jurnal"
)

# Menghapus baris template yang belum diisi
df = df.dropna(
    subset=["Kode Jurnal"]
).copy()

print("Jumlah data awal:", len(df))


# ============================================================
# 3. FILTER SPESIES
# ============================================================

# Penelitian dibatasi pada Gracilaria verrucosa

df["Jenis Gracilaria"] = (
    df["Jenis Gracilaria"]
    .astype(str)
    .str.strip()
)

df = df[
    df["Jenis Gracilaria"]
    .str.lower()
    .eq("gracilaria verrucosa")
].copy()

print("Jumlah data Gracilaria verrucosa:", len(df))

if len(df) == 0:
    raise ValueError(
        "Tidak ditemukan data Gracilaria verrucosa. "
        "Periksa kembali penulisan nama spesies di Excel."
    )


# ============================================================
# 4. PEMBERSIHAN DAN PEMERIKSAAN DATA
# ============================================================

kolom_angka = [
    "Tahun Jurnal",
    "Lama Budidaya (hari)",
    "Hari Pengamatan",
    "Berat Awal (gram)",
    "Berat Akhir (gram)",
    "Suhu Air (°C)",
    "pH Air",
    "Salinitas (ppt)",
    "Oksigen Terlarut (mg/L)",
    "Nitrat (mg/L)",
    "Fosfat (mg/L)",
    "Laju Pertumbuhan (%/hari)"
]

for kolom in kolom_angka:
    if kolom in df.columns:
        df[kolom] = pd.to_numeric(
            df[kolom],
            errors="coerce"
        )

# Menghapus duplikasi yang benar-benar identik
df = df.drop_duplicates().copy()

print("\nUkuran dataset:", df.shape)

print("\nTipe data:")
display(df.dtypes)

print("\nJumlah data kosong:")
display(df.isna().sum())

print("\nJumlah data duplikat:", df.duplicated().sum())

print("\nRingkasan data:")
display(df.describe(include="all").T)


# ============================================================
# 5. PERHITUNGAN SGR (SEBAGAI INFORMASI TAMBAHAN)
# ============================================================

# SGR tidak menjadi target utama penelitian ini.
# Perhitungan hanya dilakukan jika berat awal dan akhir
# menggunakan basis yang sama dan lama pengamatan tersedia.

syarat_sgr = (
    df["Berat Awal (gram)"].gt(0) &
    df["Berat Akhir (gram)"].gt(0) &
    df["Hari Pengamatan"].gt(0)
)

hitung_sgr = (
    syarat_sgr &
    df["Laju Pertumbuhan (%/hari)"].isna()
)

df.loc[hitung_sgr, "Laju Pertumbuhan (%/hari)"] = (
    (
        np.log(df.loc[hitung_sgr, "Berat Akhir (gram)"]) -
        np.log(df.loc[hitung_sgr, "Berat Awal (gram)"])
    ) /
    df.loc[hitung_sgr, "Hari Pengamatan"]
) * 100

print("\nData SGR sebagai informasi tambahan:")
display(
    df[
        [
            "Kode Jurnal",
            "Berat Awal (gram)",
            "Berat Akhir (gram)",
            "Hari Pengamatan",
            "Laju Pertumbuhan (%/hari)"
        ]
    ].head()
)


# ============================================================
# 6. EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# Statistik deskriptif
display(df.describe(include="all").T)

# Distribusi variabel
kolom_plot = [
    "Suhu Air (°C)",
    "pH Air",
    "Salinitas (ppt)",
    "Oksigen Terlarut (mg/L)",
    "Berat Awal (gram)",
    "Berat Akhir (gram)",
    "Laju Pertumbuhan (%/hari)"
]

tersedia = [
    kolom for kolom in kolom_plot
    if kolom in df.columns and df[kolom].notna().any()
]

for kolom in tersedia:

    plt.figure(figsize=(7, 4))

    sns.histplot(
        df[kolom].dropna(),
        kde=True
    )

    plt.title(f"Distribusi {kolom}")
    plt.xlabel(kolom)
    plt.ylabel("Frekuensi")

    plt.tight_layout()
    plt.show()


# ============================================================
# 7. ANALISIS KORELASI
# ============================================================

numerik = df.select_dtypes(include=np.number)

if numerik.shape[1] >= 2:

    plt.figure(figsize=(12, 8))

    sns.heatmap(
        numerik.corr(),
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Korelasi Antarvariabel Gracilaria verrucosa")
    plt.tight_layout()
    plt.show()

else:
    print("Data numerik belum cukup untuk analisis korelasi.")


# ============================================================
# 8. SUPERVISED LEARNING: REGRESI
# ============================================================

# Target regresi:
# Biomassa akhir (gram)

target_regresi = "Berat Akhir (gram)"

# Fitur prediktor mengikuti struktur kode awal.
# Pertambahan berat tidak dimasukkan untuk menghindari
# target leakage jika biomassa akhir dihitung dari berat awal
# + pertambahan berat.

fitur_regresi = [
    "Suhu Air (°C)",
    "pH Air",
    "Salinitas (ppt)",
    "Oksigen Terlarut (mg/L)",
    "Nitrat (mg/L)",
    "Fosfat (mg/L)",
    "Lama Budidaya (hari)",
    "Hari Pengamatan",
    "Berat Awal (gram)"
]

reg = df.dropna(
    subset=[target_regresi, "Kode Jurnal"]
).copy()

fitur_regresi = [
    kolom for kolom in fitur_regresi
    if kolom in reg.columns
    and reg[kolom].notna().any()
]

print("\nJumlah data untuk regresi:", len(reg))
print("Jumlah jurnal independen:", reg["Kode Jurnal"].nunique())
print("Fitur regresi:", fitur_regresi)


if (
    len(reg) >= 10
    and reg["Kode Jurnal"].nunique() >= 2
    and len(fitur_regresi) >= 2
):

    X = reg[fitur_regresi]
    y = reg[target_regresi]
    groups = reg["Kode Jurnal"]

    # Pembagian berdasarkan jurnal
    splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=0.25,
        random_state=42
    )

    train_idx, test_idx = next(
        splitter.split(X, y, groups=groups)
    )

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    print("\nJumlah data latih:", len(X_train))
    print("Jumlah data uji:", len(X_test))

    # Model Random Forest Regressor
    model_reg = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42
            )
        )
    ])

    model_reg.fit(X_train, y_train)

    pred_reg = model_reg.predict(X_test)

    # Evaluasi regresi
    mae = mean_absolute_error(y_test, pred_reg)

    rmse = np.sqrt(
        mean_squared_error(y_test, pred_reg)
    )

    r2 = r2_score(y_test, pred_reg)

    print("\n===== HASIL REGRESI =====")
    print("MAE :", mae)
    print("RMSE:", rmse)
    print("R2  :", r2)

    # Perbandingan nilai aktual dan prediksi
    hasil_regresi = pd.DataFrame({
        "Biomassa Aktual": y_test.values,
        "Biomassa Prediksi": pred_reg
    })

    display(hasil_regresi)

    # Grafik aktual vs prediksi
    plt.figure(figsize=(7, 5))

    plt.scatter(
        y_test,
        pred_reg
    )

    nilai_min = min(y_test.min(), pred_reg.min())
    nilai_max = max(y_test.max(), pred_reg.max())

    plt.plot(
        [nilai_min, nilai_max],
        [nilai_min, nilai_max],
        linestyle="--"
    )

    plt.xlabel("Biomassa Aktual (gram)")
    plt.ylabel("Biomassa Prediksi (gram)")
    plt.title("Random Forest Regression - Biomassa Gracilaria")

    plt.tight_layout()
    plt.show()

    # Feature importance
    rf_reg = model_reg.named_steps["model"]

    importance = pd.DataFrame({
        "Fitur": fitur_regresi,
        "Importance": rf_reg.feature_importances_
    }).sort_values(
        by="Importance",
        ascending=False
    )

    print("\nKontribusi fitur regresi:")
    display(importance)

    plt.figure(figsize=(8, 5))

    sns.barplot(
        data=importance,
        x="Importance",
        y="Fitur"
    )

    plt.title("Feature Importance Random Forest Regression")
    plt.tight_layout()
    plt.show()

else:

    print(
        "Regresi dilewati: diperlukan minimal 10 data, "
        "2 jurnal independen, dan 2 fitur yang tersedia."
    )


# ============================================================
# 9. SUPERVISED LEARNING: KLASIFIKASI
# ============================================================

# Klasifikasi tetap dipertahankan sebagai analisis eksploratif.
# Kategori dibuat berdasarkan distribusi biomassa:
# Rendah, Sedang, dan Tinggi.
#
# Kategori ini bukan standar biologis universal.

target_klasifikasi = "Berat Akhir (gram)"

klas = df.dropna(
    subset=[target_klasifikasi, "Kode Jurnal"]
).copy()

fitur_klas = [
    kolom for kolom in fitur_regresi
    if kolom in klas.columns
]

if (
    len(klas) >= 15
    and klas["Kode Jurnal"].nunique() >= 2
    and len(fitur_klas) >= 2
):

    X = klas[fitur_klas]
    y_cont = klas[target_klasifikasi]
    groups = klas["Kode Jurnal"]

    splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=0.25,
        random_state=42
    )

    train_idx, test_idx = next(
        splitter.split(X, y_cont, groups=groups)
    )

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train_cont = y_cont.iloc[train_idx]
    y_test_cont = y_cont.iloc[test_idx]

    # Ambang kelas hanya dihitung dari data latih
    batas = y_train_cont.quantile(
        [1/3, 2/3]
    ).values

    # Periksa apakah ambang batas menghasilkan 3 kelas berbeda
    if batas[0] < batas[1]:

        def buat_kategori(nilai):
            return pd.cut(
                nilai,
                bins=[
                    -np.inf,
                    batas[0],
                    batas[1],
                    np.inf
                ],
                labels=[
                    "Rendah",
                    "Sedang",
                    "Tinggi"
                ],
                include_lowest=True
            )

        y_train = buat_kategori(y_train_cont)
        y_test = buat_kategori(y_test_cont)

        if y_train.nunique() >= 2:

            # Model Random Forest Classifier
            model_klas = Pipeline([
                (
                    "imputer",
                    SimpleImputer(strategy="median")
                ),
                (
                    "scaler",
                    StandardScaler()
                ),
                (
                    "model",
                    RandomForestClassifier(
                        n_estimators=200,
                        random_state=42,
                        class_weight="balanced"
                    )
                )
            ])

            model_klas.fit(
                X_train,
                y_train
            )

            pred_klas = model_klas.predict(
                X_test
            )

            print("\n===== HASIL KLASIFIKASI =====")

            print(
                "Akurasi:",
                accuracy_score(y_test, pred_klas)
            )

            print(
                "F1 Macro:",
                f1_score(
                    y_test,
                    pred_klas,
                    average="macro",
                    zero_division=0
                )
            )

            print("\nClassification Report:")

            print(
                classification_report(
                    y_test,
                    pred_klas,
                    zero_division=0
                )
            )

            # Confusion Matrix
            label_kelas = [
                "Rendah",
                "Sedang",
                "Tinggi"
            ]

            cm = confusion_matrix(
                y_test,
                pred_klas,
                labels=label_kelas
            )

            disp = ConfusionMatrixDisplay(
                confusion_matrix=cm,
                display_labels=label_kelas
            )

            disp.plot(
                cmap="Blues",
                values_format="d"
            )

            plt.title("Confusion Matrix Random Forest Classifier")
            plt.tight_layout()
            plt.show()

            print("Batas kategori biomassa:", batas)

        else:

            print(
                "Klasifikasi dilewati: "
                "data latih hanya memiliki satu kategori."
            )

    else:

        print(
            "Klasifikasi dilewati: "
            "nilai biomassa tidak menghasilkan batas kelas yang berbeda."
        )

else:

    print(
        "Klasifikasi dilewati: data belum mencukupi "
        "untuk analisis eksploratif."
    )


# ============================================================
# 10. UNSUPERVISED LEARNING: PCA DAN K-MEANS
# ============================================================

# Analisis menggunakan variabel kualitas air.
# Tujuannya mengelompokkan kemiripan kondisi budidaya.

fitur_cluster = [
    "Suhu Air (°C)",
    "pH Air",
    "Salinitas (ppt)",
    "Oksigen Terlarut (mg/L)",
    "Nitrat (mg/L)",
    "Fosfat (mg/L)"
]

fitur_cluster = [
    kolom for kolom in fitur_cluster
    if kolom in df.columns
    and df[kolom].notna().any()
]

clu = df.dropna(
    subset=["Kode Jurnal"]
).copy()

# Buang baris yang seluruh variabel clustering-nya kosong
clu = clu.dropna(
    subset=fitur_cluster,
    how="all"
)

if (
    len(fitur_cluster) >= 2
    and len(clu) >= 6
):

    Xc = clu[fitur_cluster]

    # Imputasi dan standardisasi
    prep = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    Xc_scaled = prep.fit_transform(Xc)

    # PCA
    pca = PCA(
        n_components=2,
        random_state=42
    )

    pcs = pca.fit_transform(
        Xc_scaled
    )

    print("\n===== HASIL PCA =====")

    print(
        "Proporsi variasi komponen:",
        pca.explained_variance_ratio_
    )

    print(
        "Total variasi yang dijelaskan:",
        pca.explained_variance_ratio_.sum()
    )

    # Menentukan jumlah cluster menggunakan silhouette score
    skor_silhouette = {}

    batas_cluster = min(
        6,
        len(clu) - 1
    )

    for k in range(2, batas_cluster + 1):

        km_uji = KMeans(
            n_clusters=k,
            random_state=42,
            n_init=10
        )

        label_uji = km_uji.fit_predict(
            Xc_scaled
        )

        if len(np.unique(label_uji)) > 1:

            skor = silhouette_score(
                Xc_scaled,
                label_uji
            )

            skor_silhouette[k] = skor

    if len(skor_silhouette) > 0:

        print("\nSilhouette Score setiap cluster:")

        for k, skor in skor_silhouette.items():
            print(f"{k} Cluster: {skor:.4f}")

        # Memilih jumlah cluster dengan silhouette score tertinggi
        jumlah_cluster = max(
            skor_silhouette,
            key=skor_silhouette.get
        )

        print(
            "\nJumlah cluster terbaik:",
            jumlah_cluster
        )

        # K-Means final
        km = KMeans(
            n_clusters=jumlah_cluster,
            random_state=42,
            n_init=10
        )

        labels = km.fit_predict(
            Xc_scaled
        )

        clu["Cluster"] = labels

        # Visualisasi PCA dan K-Means
        plt.figure(figsize=(8, 6))

        plt.scatter(
            pcs[:, 0],
            pcs[:, 1],
            c=labels
        )

        plt.xlabel("Komponen Utama 1")
        plt.ylabel("Komponen Utama 2")

        plt.title(
            "PCA dan K-Means Gracilaria verrucosa"
        )

        plt.tight_layout()
        plt.show()

        print("\nHasil pengelompokan:")
        display(
            clu[
                ["Kode Jurnal", "Cluster"] +
                fitur_cluster
            ]
        )

    else:

        print(
            "K-Means dilewati: "
            "cluster tidak dapat dibentuk secara valid."
        )

else:

    print(
        "PCA/K-Means dilewati: diperlukan minimal "
        "2 fitur kualitas air dan 6 baris data."
    )


# ============================================================
# 11. MENYIMPAN DATA HASIL PEMBERSIHAN
# ============================================================

file_output = "Data_Gracilaria_verrucosa_Bersih.xlsx"

df.to_excel(
    file_output,
    index=False
)

print("\nDataset hasil pembersihan berhasil disimpan:")
print(file_output)


# ============================================================
# 12. CATATAN INTERPRETASI
# ============================================================

print("""
CATATAN ANALISIS:

1. Regresi digunakan untuk memperkirakan biomassa akhir
   Gracilaria verrucosa.

2. Klasifikasi biomassa rendah, sedang, dan tinggi
   merupakan pengelompokan relatif berdasarkan dataset,
   bukan standar biologis universal.

3. PCA dan K-Means digunakan untuk melihat pola kemiripan
   kondisi kualitas air selama budidaya.

4. Data dari satu jurnal tidak boleh tersebar ke data
   latih dan data uji.

5. Biomassa basah dan kering tidak boleh digabungkan
   tanpa dasar konversi yang valid.

6. Jika jumlah jurnal independen masih sedikit,
   hasil model hanya bersifat eksploratif dan belum
   dapat digeneralisasikan.
""")


hasil_klasifikasi = pd.DataFrame({
    "Aktual": y_test.values,
    "Prediksi": pred_klas
})

print("Perbandingan kategori aktual dan prediksi:")
display(hasil_klasifikasi)

print("\nConfusion Matrix:")
from sklearn.metrics import confusion_matrix

print(confusion_matrix(
    y_test,
    pred_klas,
    labels=["Rendah", "Sedang", "Tinggi"]
))
```
